## Pandas is powerful but hard to use

Pandas is the most popular Python library for data analysis. While it offers a great deal of functionality, it is also considered a fairly difficult library to learn well. Some reasons for this include:

- There are often multiple ways to accomplish common tasks.
- There are more than 240 DataFrame attributes and methods.
- Several methods are aliases of one another (they reference the exact same underlying code).
- Several methods have nearly identical functionality.
- There are many tutorials written by different people showing different ways to do the same thing.
- There is no official document with guidelines on how to idiomatically complete common tasks.
- The official documentation itself contains non-idiomatic code.

## Problems with dot notation

There are three problems with using dot notation. It does not work in the following situations:

- When there are spaces in the column name
- When the column name is the same as a DataFrame method
- When the column name is a variable

If the desired column name has spaces, you will not be able to select it with dot notation. Python uses spaces to separate names and operators and therefore will not treat a column name containing a space as correct syntax. Let's trigger this error.

## Changing the column order of a Pandas DataFrame

During the data-preprocessing and feature-creation stage, it can happen that you end up with columns that are not necessarily in the order you would like. In this tutorial we will look at how to change the column order of a pandas DataFrame.

## How do you sort a pandas DataFrame?

You can use the pandas DataFrame `sort_values()` function to sort a DataFrame. It offers the flexibility to sort a DataFrame by one or more columns, choose the sorting algorithm, decide how to treat NaNs during comparisons, use a custom key for sorting, etc. The syntax is:

```python
df.sort_values(by, ascending=True, inplace=True)
```

Pass the column or list of columns to sort by to the `by` parameter. By default it returns a sorted DataFrame and does not modify the original DataFrame. If you want to modify the original DataFrame in place, pass `inplace=True`.

We will continue with the DataFrame defined earlier.

## 1. Sort a DataFrame by a single column

To sort a DataFrame by a single column, pass the column name to the `by` parameter of the `sort_values()` function. For example, to sort the DataFrame above by `Shares`:

````python
df_sorted = df.sort_values(by='Shares',ascending=False)
print(df_sorted)
````

````python
df
````

## 2. Sort a DataFrame by multiple columns

You can also sort a pandas DataFrame by several columns. For this, pass the columns you want to sort the DataFrame by as a list to the `by` parameter. For example, to sort the DataFrame `df` by `Shares` and `Symbol`:

````python
df_sorted = df.sort_values(by=['Shares','Symbol'],ascending=[False,True])
print(df_sorted)
````

## 3. Sort the DataFrame placing NaNs first

If you want to sort a DataFrame by a column that has one or more NaN values, you can decide whether to place them first or last. This behavior is controlled by the `na_position` parameter, which takes the values `'first'` and `'last'` and is `'last'` by default. Here is an example:

First, we create a copy of the original DataFrame and set Apple's shares to NaN.

````python
import numpy as np
df_na = df.copy()
df_na.loc[3, 'Shares'] = np.nan
print(df_na)
````

Then we sort the DataFrame `df_na` by the `Shares` column so that the NaNs are placed first in the ordering.

````python
df_na[df_na.isna()]=np.mean(df_na.Shares)
````

````python
df_na
````

## 4. Sort the DataFrame with a different sorting algorithm

The `sort_values()` function also lets you choose among three different sorting algorithms. The `kind` parameter controls this behavior. It takes `'quicksort'`, `'mergesort'`, and `'heapsort'` as values and is `'quicksort'` by default. For example, to sort a DataFrame by a column using `'mergesort'` as the sorting algorithm:

````python
df_sorted = df.sort_values(by='Shares', kind='mergesort')
print(df_sorted)
````

## Pandas DataFrame to a Python list

Pandas DataFrames are great for manipulating data. But sometimes you may prefer to have the data as a list (or more precisely, a list of lists). In this tutorial we will look at how to convert a pandas DataFrame into a Python list.

## How do you convert a pandas DataFrame into a list?

There are several ways to get a Python list from a pandas DataFrame, depending on the kind of list you want to create. To quickly get a list from a DataFrame with each element representing a row in the DataFrame, you can use the `tolist()` function as `df.values.tolist()`.

However, there are other ways too. You can create a list with each element representing a DataFrame column. Or you can create something very specific based on your requirements. Let's look at some of the different use cases with examples.

## Examples
Continuing with our DataFrame.

### 1. List with DataFrame rows as elements

As mentioned above, you can quickly get a list from a DataFrame using the `tolist()` function.

````python
ls = df.values.tolist()
print(ls)
````

### 2. List with DataFrame columns as elements
You can also use the `tolist()` function on individual columns of a DataFrame to get a list of column values.

````python
# list with each item representing a column
ls = []
for col in df.columns:
    # convert pandas series to list
    col_ls = df[col].tolist()
    # append column list to ls
    ls.append(col_ls)
# print the created list
print(ls)
````

In the example above, we iterate through each column of the DataFrame, convert it to a list, and then append it to `ls`. You can see that we get a list of lists, with each element of the list representing a column in the DataFrame.

### 3. List from a DataFrame by iterating through the rows
You can also create a list by iterating through the rows of the DataFrame.

````python
ls = []
# iterate over the rows
for i, row in df.iterrows():
    # create a list representing the dataframe row
    row_ls = [row['Name'], row['Industry'],row['Shares'], row['Symbol']]
    # append row list to ls
    ls.append(row_ls)

print(ls)
````

In the example above, we use the pandas DataFrame `iterrows()` function to iterate over the rows of `df` and create a list with row values that is appended to `ls`. You can see that we get a list of lists, with each element representing a row in the DataFrame, just as we saw in the example with the `tolist()` function.

This method also gives you the flexibility to create specific lists according to your requirements. For example, from the DataFrame above, if you want to create a list of lists with only the stock symbol and its respective share count, you can easily do so by keeping only those fields.

````python
ls = []
# iterate over the rows
for i, row in df.iterrows():
    # create a list representing the dataframe row
    row_ls = [row['Symbol'], row['Shares']]
    # append row list to ls
    ls.append(row_ls)

print(ls)
````

The following are some of the ways to get a list from a pandas DataFrame, explained with examples.

As mentioned above, you can quickly get a list from a DataFrame using the `tolist()` function.

````python
ls = df.values.tolist()
print(ls)
````

## Counting unique values in each column

Generally, the data in each column represents a different feature of a pandas DataFrame. It can be continuous, categorical, or something entirely different such as distinct texts. If you are not sure about the nature of the values you are dealing with, it could be a good exploratory step to know the count of distinct values. In this tutorial we will look at how to get the count of unique values in each column of a pandas DataFrame.

### The `nunique()` function
To count the unique values of each column of a DataFrame, you can use the pandas DataFrame `nunique()` function. The syntax is:

```python
counts = df.nunique()
```

Here, `df` is the DataFrame for which you want to know the unique counts. It returns a pandas Series of counts. By default, the pandas DataFrame `nunique()` function counts the distinct values along `axis=0`, that is, across rows, which gives you the count of distinct values in each column.

### Examples

Let's look at some of the different use cases for getting unique counts through a few examples. First, we will create a sample DataFrame that we will use throughout this tutorial.

````python
import pandas as pd
import numpy as np
# create a sample dataframe
data = {
    'EmpCode': ['E1', 'E2', 'E3', 'E4', 'E5'],
    'Gender': ['Male', 'Female', 'Female', 'Male', 'Male'],
    'Age': [27, 24, 29, 24, 25],
    'Department': ['Accounting', 'Sales', 'Accounting', np.nan, 'Sales']
}
df = pd.DataFrame(data)
# display the dataframe
df
````

Here, we create a DataFrame with information about some employees in an office. The DataFrame has the following columns: `EmpCode`, `Gender`, `Age`, and `Department`.

### 1. Count of unique values in each column

Using the pandas DataFrame `nunique()` function with default parameters gives a count of all distinct values in each column.

````python
print(df.nunique())
````

In the example above, the `nunique()` function returns a pandas Series with the counts of distinct values in each column. Note that for the `Department` column we only have two distinct values, since the `nunique()` function, by default, ignores all NaN values.

### 2. Count of unique values in each row
You can also get the count of distinct values in each row by setting the `axis` parameter to 1 or `'columns'` in the `nunique()` function.

````python
# count of unique values in each row
print(df.nunique(axis=1))
````

In the example above, you can see that we have 4 distinct values in each row except the row with index 3, which has 3 unique values due to the presence of a NaN value.

### Good to know

In case you want to know the count of each of the distinct values of a specific column, you can use the pandas `value_counts()` function. In the DataFrame `df` above, if you want to know the count of each distinct value in the `Gender` column, you can use:

````python
# count of each unique value in the "Gender" column
print(df['Gender'].value_counts())
````

In the example above, the pandas Series `value_counts()` function is used to get the counts of `'Male'` and `'Female'`, the distinct values in the `Gender` column of the DataFrame `df`.

## Removing duplicates from a Pandas DataFrame

While working with data, there may be situations where your DataFrame has duplicate rows. Knowing how to remove such rows quickly can be very useful. In this tutorial we will look at how to remove duplicates from a pandas DataFrame through a few examples.

### The `drop_duplicates()` function
The pandas DataFrame `drop_duplicates()` function can be used to remove duplicate rows from a DataFrame. It also gives you the flexibility to identify duplicates based on certain columns through the `subset` parameter. Its syntax is:

```python
df.drop_duplicates()
```

It returns a DataFrame with the duplicate rows removed. By default it removes duplicates except for the first occurrence. You can change this behavior through the `keep` parameter, which takes `'first'`, `'last'`, or `False`. To modify the DataFrame in place, pass the argument `inplace=True`.

### Examples

Let's look at some of the use cases of the `drop_duplicates()` function through examples:

### 1. Remove duplicate rows based on all columns
By default, the `drop_duplicates()` function identifies duplicates taking all columns into account. It then removes the duplicate rows and keeps only their first occurrence.

````python
import pandas as pd

# create a sample dataframe with duplicate rows
data = {
    'Pet': ['Cat', 'Dog', 'Dog', 'Dog', 'Cat'],
    'Color': ['Brown', 'Golden', 'Golden', 'Golden', 'Black'],
    'Eyes': ['Black', 'Black', 'Black', 'Brown', 'Green']
}

df = pd.DataFrame(data)

# print the dataframe
print("The original dataframe:\n")
print(df)

# drop duplicates
df_unique = df.drop_duplicates()
print("\nAfter dropping duplicates:\n")
print(df_unique)
````

In the example above, you can see that the rows with index 1 and 2 have the same values for all three columns. By applying the `drop_duplicates()` function, the first row is kept and the remaining duplicate rows are discarded. As a result, the returned DataFrame does not have a continuous index. If you want the returned DataFrame to have a continuous index, pass `ignore_index=True` to the `drop_duplicates()` function or reset the index of the returned DataFrame.

### 2. Drop duplicate rows based on certain columns

You can also tell the `drop_duplicates()` function to identify duplicates based only on certain columns by passing them as a list to the `subset` argument.

````python
import pandas as pd

# create a sample dataframe with duplicate rows
data = {
    'Pet': ['Cat', 'Dog', 'Dog', 'Dog', 'Cat'],
    'Color': ['Brown', 'Golden', 'Golden', 'Golden', 'Black'],
    'Eyes': ['Black', 'Black', 'Black', 'Brown', 'Green']
}

df = pd.DataFrame(data)

# print the dataframe
print("The original dataframe:\n")
print(df)

# drop duplicates
df_unique = df.drop_duplicates(subset=['Pet', 'Color'])
print("\nAfter dropping duplicates:\n")
print(df_unique)
````

In the example above, we identify duplicates based only on the `Pet` and `Color` columns by passing them as a list to the `drop_duplicates()` function. With this criterion, the rows with index 1, 2, and 3 are now duplicates and the returned DataFrame keeps only the first row.

### 3. Drop duplicates and keep the last occurrence

If you want to keep the last duplicate row instead of the first, pass `keep='last'` to the `drop_duplicates()` function.

````python
import pandas as pd

# create a sample dataframe with duplicate rows
data = {
    'Pet': ['Cat', 'Dog', 'Dog', 'Dog', 'Cat'],
    'Color': ['Brown', 'Golden', 'Golden', 'Golden', 'Black'],
    'Eyes': ['Black', 'Black', 'Black', 'Brown', 'Green']
}

df = pd.DataFrame(data)

# print the dataframe
print("The original dataframe:\n")
print(df)

# drop duplicates
df_unique = df.drop_duplicates(keep='last')
print("\nAfter dropping duplicates:\n")
print(df_unique)
````

In the example above, we keep the last duplicate instead of the first.

### 4. Drop duplicates and keep no occurrence
If you do not want to keep any of the duplicate rows, pass `keep=False` to the `drop_duplicates()` function.

````python
import pandas as pd

# create a sample dataframe with duplicate rows
data = {
    'Pet': ['Cat', 'Dog', 'Dog', 'Dog', 'Cat'],
    'Color': ['Brown', 'Golden', 'Golden', 'Golden', 'Black'],
    'Eyes': ['Black', 'Black', 'Black', 'Brown', 'Green']
}

df = pd.DataFrame(data)

# print the dataframe
print("The original dataframe:\n")
print(df)

# drop duplicates
df_unique = df.drop_duplicates(keep=False)
print("\nAfter dropping duplicates:\n")
print(df_unique)
````

## Concatenating DataFrames in Pandas

You can concatenate DataFrames along the row or the column axis. Suppose you have several DataFrames with the same fields and you would like to combine them into one. In this case you will want to concatenate the DataFrames along the index or row axis. Or suppose you have some extra fields for your current data that you want to add — then you will want to concatenate them along the column axis. In this tutorial we will look at how to concatenate two or more DataFrames in pandas.

### The pandas `concat()` function
The pandas `concat()` function is used to concatenate multiple DataFrames into one. Its syntax is:

```python
pd.concat(objs, axis=0)
```

Pass the sequence of DataFrame objects (`objs`) you want to concatenate and indicate the axis (0 for rows and 1 for columns) along which the concatenation will be performed; it returns the concatenated DataFrame. The default value for the axis is 0. Note that you can also use the `concat` function to concatenate pandas Series.

### Examples
Let's look at some of the use cases of the `concat()` function through examples:

### 1. Concatenate DataFrames with the same fields (vertically)
If you have two or more DataFrames that have the same fields (or columns) and you want to concatenate them into one:

````python
import pandas as pd

# Create two dataframes with same columns
df1 = pd.DataFrame({'Name': ['Sam', 'Emma'], 'Age': [14, 15]})
df2 = pd.DataFrame({'Name': ['Karen', 'Rahul'], 'Age': [10, 13]})

# Print the dataframes
print("DataFrame df1:\n")
print(df1)
print("\nDataFrame df2:\n")
print(df2)

# Concatenate them along the axis
df_combined = pd.concat([df1, df2])

# Print the concatenated dataframe
print("\nCombined DataFrame:\n", df_combined)
````

In the above example, the dataframes df1 and df2 have the same columns – Name and Age and are concatenated along the index, that is vertically (since the axis parameter is by default 0). Also, note that the combined dataframe retains the index from the individual dataframes, you can reset the index by using the reset_index() function.

````python
df_combined = df_combined.reset_index(drop=True)
print(df_combined)
````

The above is the combined DataFrame with its index reset.

### 2. Concatenate DataFrames with rows referring to the same records (horizontally)

If you have two or more DataFrames with rows that refer to the same records and you want to concatenate them so that all the different fields are in a single DataFrame, use the `concat()` function with `axis=1`.

````python
import pandas as pd

# Create two dataframes with same columns
df1 = pd.DataFrame({'Name': ['Sam', 'Emma'], 'Age': [14, 15]})
df2 = pd.DataFrame({'Math': ['B', 'A+'], 'Science': ['A', 'B+']})

# Print the dataframes
print("DataFrame df1:\n")
print(df1)
print("\nDataFrame df2:\n")
print(df2)

# Concatenate them along the axis
df_combined = pd.concat([df1, df2], axis=1)

# Print the concatenated dataframe
print("\nCombined DataFrame:\n", df_combined)
````

In the example above, the DataFrames df1 (storing the students' name and age) and df2 (storing their Math and Science grades) are concatenated horizontally (along the column axis).

## Dropping one or more columns from a DataFrame

Pandas DataFrames are quite powerful for manipulating data. Often, while working with data — particularly during EDA (Exploratory Data Analysis) and data preprocessing — you may need to remove one or more columns. In this tutorial we will look at how to drop one or more columns from a pandas DataFrame with some examples.

### How do you drop columns from a pandas DataFrame?

You can use the pandas DataFrame `drop()` function with the axis set to 1 to remove one or more columns from a DataFrame. The syntax is:

```python
df.drop(cols_to_drop, axis=1)
```

Here, `cols_to_drop` is the index or labels of the columns to drop; if more than one column is to be dropped, it should be a list. The axis represents the axis from which to drop the labels; the default is 0, but if you want to drop columns, pass the axis as 1 (that is, 0 for rows and 1 for columns).

Also note that the `drop()` function does not modify the DataFrame in place by default. It returns a copy of the DataFrame with the labels removed. If you want to modify the DataFrame in place, pass the argument `inplace=True` to the function.

### Drop columns by name
To drop columns by name, simply pass the column name (if you want to drop a single column) or the list of columns (if you want to drop several columns) to the `drop` function. See the examples below:

### Example 1: drop a single column by name

````python
import pandas as pd

# create a sample dataframe
data = {
    'A': ['a1', 'a2', 'a3'],
    'B': ['b1', 'b2', 'b3'],
    'C': ['c1', 'c2', 'c3'],
    'D': ['d1', 'd2', 'd3']
}

df = pd.DataFrame(data)

# print the dataframe
print("Original Dataframe:\n")
print(df)

# remove column C
df = df.drop('C', axis=1)

print("\nAfter dropping C:\n")
print(df)
````

In the example above, a sample DataFrame `df` is created with four columns A, B, C, and D. Then column C is dropped using the `drop()` function. Note that, since we only had to drop one column, we did not need to pass a list.

### Example 2: drop several columns by name

````python
import pandas as pd

# create a sample dataframe
data = {
    'A': ['a1', 'a2', 'a3'],
    'B': ['b1', 'b2', 'b3'],
    'C': ['c1', 'c2', 'c3'],
    'D': ['d1', 'd2', 'd3']
}

df = pd.DataFrame(data)

# print the dataframe
print("Original Dataframe:\n")
print(df)

# remove columns C and D
df = df.drop(['C', 'D'], axis=1)

print("\nAfter dropping columns C and D:\n")
print(df)
````

In the example above, columns C and D are dropped from the DataFrame `df`. Note that we had to provide the list of column names to drop since we were dropping several columns together.

### Drop columns by index

To drop columns by column number, pass `df.columns[i]` to the `drop()` function, where `i` is the column index of the column you want to drop. To drop several columns by their indices, pass `df.columns[[i, j, k]]`, where `i, j, k` are the column indices of the columns you want to drop.

### Example 1: drop a single column by index

````python
import pandas as pd

# create a sample dataframe
data = {
    'A': ['a1', 'a2', 'a3'],
    'B': ['b1', 'b2', 'b3'],
    'C': ['c1', 'c2', 'c3'],
    'D': ['d1', 'd2', 'd3']
}

df = pd.DataFrame(data)

# print the dataframe
print("Original Dataframe:\n")
print(df)

# remove column
df = df.drop(df.columns[2], axis=1)

print("\nAfter dropping C:\n")
print(df)
````

In the example above, column C is dropped using its index 2 from the DataFrame `df`.

### Example 2: drop several columns by their index

````python
import pandas as pd

# create a sample dataframe
data = {
    'A': ['a1', 'a2', 'a3'],
    'B': ['b1', 'b2', 'b3'],
    'C': ['c1', 'c2', 'c3'],
    'D': ['d1', 'd2', 'd3']
}

df = pd.DataFrame(data)

# print the dataframe
print("Original Dataframe:\n")
print(df)

# remove columns C and D
df = df.drop(df.columns[[2, 3]], axis=1)

print("\nAfter dropping columns C and D:\n")
print(df)
````

In the example above, the columns with index 2 and 3 are dropped from the DataFrame `df`.

Behind the scenes, `df.columns` actually gives the column name for the given index:

## Rolling windows on a time series

The following examples read an hourly energy-demand series, index it by date, and compute moving (rolling) aggregations — a common operation when smoothing or summarizing time-series data.

> Note: these cells expect the original data files (e.g. `Data/PJM/PJMW_hourly.csv`, `Data/Energy/...`). Point the paths at your own CSVs to run them.
